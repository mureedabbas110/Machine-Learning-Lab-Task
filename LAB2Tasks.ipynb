````python
#1.	Line Graph — Plot a week of your own made-up daily step counts as a line graph with markers, a title, and axis labels.
#```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]
step_counts = [1000, 1500, 2000, 1800, 2200, 2500, 3000]

plt.plot(days, step_counts, marker='o')
plt.title('Daily Step Counts')
plt.xlabel('Days')
plt.ylabel('Step Count')
plt.show()
````

````python
import matplotlib.pyplot as plt
hours_sleep = [7, 6, 8, 5, 7, 6, 9]
productivity = [5, 4, 6, 3, 5, 4, 7]
plt.scatter(hours_sleep, productivity , color='blue', marker='o')
plt.title('Hours of Sleep vs Productivity')
plt.xlabel('Hours of Sleep')
plt.ylabel('Productivity')
plt.show()
````

````python
Products = ['A', 'B', 'C', 'D', 'E']
sales_q1 = [100, 150, 200, 250, 300]
sales_q2 = [120, 180, 220, 270, 320]
import numpy as np
x = np.arange(len(Products))
width = 0.35
plt.bar(x - width/2, sales_q1, width, label='Q1')
plt.bar(x + width/2, sales_q2, width, label='Q2')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.title('Sales Comparison')
plt.xticks(x, Products)
plt.legend()
plt.show()
````

````python
import matplotlib.pyplot as plt
goroup_1 = [10, 15, 20, 25, 30]
group_2 = [12, 18, 22, 28, 35]
group_3 = [8, 14, 18, 24, 32]

plt.boxplot([goroup_1, group_2, group_3], label=['Group 1', 'Group 2', 'Group 3'])
plt.title('Box Plot of Three Groups')   
plt.ylabel('Values')
plt.show()
````

````python
#Pie Chart — Create a pie chart of your own weekly time allocation across at least 4 categories (must sum to 100%), with one slice exploded.
import matplotlib.pyplot as plt
categories = ['Work', 'Exercise', 'Leisure', 'Sleep']
time_allocation = [40, 10, 20, 30]  
plt.pie(time_allocation, labels=categories, autopct='%1.1f%%', startangle=90, explode=(0.1, 0, 0, 0))
plt.title('Weekly Time Allocation')
plt.show()
````

````python

#Capstone Challenge — Choose any built-in dataset (e.g., load_wine or load_breast_cancer from sklearn.datasets), build a 2x3 subplot 'Visual EDA Report' like Task 8 using at least four different chart types, and write a short written summary (4–5 sentences) of what the visualizations reveal about the dataset
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine

wine_data = load_wine()
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
# Histogram of alcohol content
axes[0, 0].hist(wine_data.data[:, 0], bins=20, color='skyblue', edgecolor='black')
axes[0, 0].set_title('Histogram of Alcohol Content')
# Scatter plot of alcohol vs malic acid
axes[0, 1].scatter(wine_data.data[:, 0], wine_data.data[:, 1], color='green')
axes[0, 1].set_title('Alcohol vs Malic Acid')
# Box plot of color intensity
axes[0, 2].boxplot(wine_data.data[:, 9])
axes[0, 2].set_title('Box Plot of Color Intensity')
# Bar chart of wine classes
classes, counts = np.unique(wine_data.target, return_counts=True)
axes[1, 0].bar(classes, counts, color='orange')
axes[1, 0].set_title('Bar Chart of Wine Classes')
# Pie chart of wine classes
axes[1, 1].pie(counts, labels=classes, autopct='%1.1f%%', startangle=90)
axes[1, 1].set_title('Pie Chart of Wine Classes')
# Line plot of flavanoids
axes[1, 2].plot(wine_data.data[:, 7], color='purple')
axes[1, 2].set_title('Line Plot of Flavanoids')












````
