```python
#Task One
import pandas as pd
import numpy as np

# Create dataset with missing values
df = pd.DataFrame({
    "Age": [20, 22, np.nan, 25, 24, np.nan, 28, 30],
    "Income": [25000, np.nan, 30000, 35000, np.nan, 42000, 45000, 50000]
})

# Before imputation
print("BEFORE IMPUTATION")
print(df)

print("\nSummary Statistics Before Imputation:")
print(df.describe())

# Mean imputation for Age
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Median imputation for Income
df["Income"] = df["Income"].fillna(df["Income"].median())

# After imputation
print("\nAFTER IMPUTATION")
print(df)

print("\nSummary Statistics After Imputation:")
print(df.describe())
```

```python
#Task Two
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Create dataset
df = pd.DataFrame({
    "Size": ["S", "M", "L", "XL", "M", "S", "L", "XL"],
    "Favorite_Color": [
        "Red", "Blue", "Green", "Red",
        "Yellow", "Blue", "Green", "Black"
    ]
})

print("ORIGINAL DATASET")
print(df)

# Ordinal Encoding for Size
# S < M < L < XL
ordinal_encoder = OrdinalEncoder(
    categories=[["S", "M", "L", "XL"]]
)

df["Size_Encoded"] = ordinal_encoder.fit_transform(
    df[["Size"]]
).astype(int)

# One-Hot Encoding for Favorite Color
one_hot = pd.get_dummies(
    df["Favorite_Color"],
    prefix="Color",
    dtype=int
)

# Combine encoded columns
df_encoded = pd.concat([df, one_hot], axis=1)

print("\nENCODED DATASET")
print(df_encoded)
```

### Explanation

Size is an ordinal categorical variable because the categories have a natural order:
S < M < L < XL. Therefore, ordinal encoding is appropriate.

Favorite Color is a nominal categorical variable because colors do not have
any natural order. Therefore, one-hot encoding is appropriate.

```python
#Task Three
import pandas as pd
from sklearn.model_selection import train_test_split

# Create dataset
X = pd.DataFrame({
    "Feature1": range(100),
    "Feature2": range(100, 200)
})

# Imbalanced target:
# 80% Class 0
# 20% Class 1
y = pd.Series(
    [0] * 80 + [1] * 20,
    name="Target"
)

# Split WITHOUT stratify
X_train1, X_test1, y_train1, y_test1 = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Split WITH stratify
X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Original proportions
print("ORIGINAL CLASS PROPORTIONS")
print(y.value_counts(normalize=True))

# Without stratification
print("\nWITHOUT STRATIFY")
print(y_test1.value_counts(normalize=True))

# With stratification
print("\nWITH STRATIFY")
print(y_test2.value_counts(normalize=True))
```

### Explanation

Stratification is useful when the target variable is imbalanced. Using
stratify=y keeps approximately the same class proportions in the training
and testing datasets as in the original dataset.

```python
#Task Four
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Create dataset
df = pd.DataFrame({
    "Age": [18, 22, 25, 30, 35, 40, 45, 50],
    "Income": [
        20000, 25000, 30000, 40000,
        50000, 65000, 80000, 100000
    ]
})

print("ORIGINAL DATA")
print(df)

# Function to display statistics
def show_statistics(data, title):
    print("\n" + title)

    print("\nMinimum:")
    print(data.min())

    print("\nMaximum:")
    print(data.max())

    print("\nMean:")
    print(data.mean())

    print("\nStandard Deviation:")
    print(data.std())


# Before scaling
show_statistics(
    df,
    "STATISTICS BEFORE SCALING"
)

# MinMaxScaler
minmax_scaler = MinMaxScaler()

df_minmax = pd.DataFrame(
    minmax_scaler.fit_transform(df),
    columns=df.columns
)

show_statistics(
    df_minmax,
    "STATISTICS AFTER MINMAX SCALER"
)

# StandardScaler
standard_scaler = StandardScaler()

df_standard = pd.DataFrame(
    standard_scaler.fit_transform(df),
    columns=df.columns
)

show_statistics(
    df_standard,
    "STATISTICS AFTER STANDARD SCALER"
)
```

### Explanation

MinMaxScaler converts the values of each feature to a common range, normally
between 0 and 1.

StandardScaler transforms the data so that each feature has approximately
mean 0 and standard deviation 1.

```python
#Task Five
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

# Load Iris dataset
iris = load_iris()

# Features
X = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Target
y = iris.target

print("FEATURES")
print(X.head())

# Create Random Forest model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train model
model.fit(X, y)

# Calculate feature importance
importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

# Sort by importance
importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

print("\nFEATURE IMPORTANCE RANKING")
print(importance_df)

# Top 3 features
print("\nTOP 3 FEATURES")
print(importance_df.head(3))
```

### Explanation

Random Forest provides an importance score for each feature. A higher
importance score means that the feature contributed more to the model's
predictions.

The features are sorted from highest to lowest importance, and the top
three features are selected.

```python
#Task Six
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer


# ==========================================
# CREATE DATASET
# ==========================================

df = pd.DataFrame({
    "Age": [
        20, 22, np.nan, 25, 28,
        30, np.nan, 35, 40, 45
    ],

    "Income": [
        25000, 30000, 35000, np.nan, 45000,
        50000, 55000, np.nan, 70000, 80000
    ],

    "City": [
        "Khairpur", "Sukkur", "Khairpur",
        "Larkana", "Sukkur", "Khairpur",
        "Larkana", "Sukkur", "Khairpur",
        "Larkana"
    ],

    "Purchased": [
        0, 1, 0, 1, 1,
        0, 1, 1, 0, 1
    ]
})


print("======================================")
print("ORIGINAL DATASET")
print("======================================")

print(df)
print("\nShape:", df.shape)


# ==========================================
# STEP 1 — HANDLE MISSING DATA
# ==========================================

numeric_columns = ["Age", "Income"]

imputer = SimpleImputer(
    strategy="median"
)

df[numeric_columns] = imputer.fit_transform(
    df[numeric_columns]
)


print("\n======================================")
print("STEP 1 — AFTER MISSING DATA HANDLING")
print("======================================")

print(df)
print("\nShape:", df.shape)


# ==========================================
# STEP 2 — ENCODE CATEGORICAL DATA
# ==========================================

encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

encoded_city = encoder.fit_transform(
    df[["City"]]
)

encoded_city_df = pd.DataFrame(
    encoded_city,
    columns=encoder.get_feature_names_out(["City"]),
    index=df.index
)

df_encoded = pd.concat(
    [
        df.drop(columns=["City"]),
        encoded_city_df
    ],
    axis=1
)


print("\n======================================")
print("STEP 2 — AFTER CATEGORICAL ENCODING")
print("======================================")

print(df_encoded.head())
print("\nShape:", df_encoded.shape)


# ==========================================
# STEP 3 — TRAIN/TEST SPLIT
# ==========================================

X = df_encoded.drop(
    columns=["Purchased"]
)

y = df_encoded["Purchased"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\n======================================")
print("STEP 3 — TRAIN/TEST SPLIT")
print("======================================")

print("X_train Shape:", X_train.shape)
print("X_test Shape:", X_test.shape)

print("\nX_train Preview:")
print(X_train.head())

print("\nX_test Preview:")
print(X_test.head())


# ==========================================
# STEP 4 — FEATURE SCALING
# ==========================================

scaler = StandardScaler()

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

scale_columns = [
    "Age",
    "Income"
]

# Fit only on training data
X_train_scaled[scale_columns] = scaler.fit_transform(
    X_train[scale_columns]
)

# Transform test data using the same scaler
X_test_scaled[scale_columns] = scaler.transform(
    X_test[scale_columns]
)


print("\n======================================")
print("STEP 4 — AFTER FEATURE SCALING")
print("======================================")

print("X_train_scaled Shape:", X_train_scaled.shape)
print("X_test_scaled Shape:", X_test_scaled.shape)

print("\nScaled Training Data:")
print(X_train_scaled.head())

print("\nScaled Testing Data:")
print(X_test_scaled.head())
```

### Mini-Pipeline Explanation

The preprocessing steps are performed in the required order.

1. Missing values in Age and Income are handled using median imputation.
2. The City categorical column is converted into numerical one-hot encoded columns.
3. The dataset is divided into training and testing sets using stratification.
4. Age and Income are standardized using StandardScaler.

The scaler is fitted only on the training data and then used to transform
the test data. This prevents information from the test set from leaking
into the training process.
